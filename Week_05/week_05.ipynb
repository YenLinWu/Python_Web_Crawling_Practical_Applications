#### 字串方法為什麼不夠用?

```python
s1 = "In stock 22 available"
s2 = "22 available"          
 
print(s1.split(" ")[2])   # '22'
print(s2.split(" ")[2])   # IndexError
```

#### **正規表達式**

```python
import re

print(re.search(r"\d+", s1).group())   # '22'
print(re.search(r"\d+", s2).group())   # '22'
```

#### **字元類**


```python
import re

string = "tom.wu @ Tainan @ 1112"
 
print(re.findall(r"[tom]", string))   # ['t', 'o', 'm']
print(re.findall(r"[0-9]", string))   # ['1', '1', '1', '2']
print(re.findall(r"\s", string))      # [' ', ' ', ' ', ' ']
print(re.findall(r"\W", string))      # ['.', ' ', '@', ' ', ' ', '@', ' ']
print(re.findall(r"\w", string))      # ['t', 'o', 'm', 'w', 'u', 'T', 'a', 'i', 'n', 'a', 'n', '1', '1', '1', '2']
```

#### **量詞**

```python
import re

string = "tom.wu @ Tainan @ 1112"
 
print(re.findall(r"[tom]+", string))   # ['tom']
print(re.findall(r"[0-9]+", string))      # ['1112']
print(re.findall(r"\w+", string))          # ['tom', 'wu', 'Tainan', '1112']
```

```python
import re

dates = """2026-10-19
        2026-1-9
        20261019
        115-10-19
        2026 - 10 - 19"""

print(re.findall(r"\d{4}-\d{2}-\d{2}", dates))       # ['2026-10-19']
print(re.findall(r"\d{4}-\d{1,2}-\d{1,2}", dates))   # ['2026-10-19', '2026-1-9']
print(re.findall(r"\d{4}-?\d{2}-?\d{2}", dates))     # ['2026-10-19', '20261019']
print(re.findall(r"\d+-\d+-\d+", dates))             # ['2026-10-19', '2026-1-9', '115-10-19']
```

#### **中介字元**

```python
import re

string = "tom.wu @ Tainan @ 1112"   

print(re.findall(r"\w+", string))        # ['tom', 'wu', 'Tainan', '1112']
print(re.findall(r"^\w+", string))       # ['tom']
print(re.findall(r" \w+$", string))      # ['1112']

print(re.findall(r"\d{2}+", string))      # ['11', '12']
print(re.findall(r"\b\d{4}\b", string))   # ['1112']    

print(re.findall(r"\w+.\w+", string))     # ['tom.wu', 'Tainan', '1112']
print(re.findall(r"\w+\.\w+", string))    # ['tom.wu']
```

#### **re 模組常用的函式**

```python
import re

string = "tom.wu @ Tainan @ 1112"
 
print(re.findall(r"\w+", string))            # ['tom', 'wu', 'Tainan', '1112']
 
print(re.search(r"\d+", string).group())     # '1112'  第一個數字串
print(re.match(r"\w+", string).group())      # 'tom'   開頭為字串
# print(re.match(r"\d+", string).group())      # None  開頭非數字
 

# 月份英文 + 空格 + 日 + 逗號空格 + 4 位數年份
DATE = r"[A-Z][a-z]+ \d{1,2}, \d{4}"

print(re.fullmatch(DATE, "March 14, 1879"))    # Match   格式正確
print(re.fullmatch(DATE, "June 01, 1926"))     # Match   格式正確
# print(re.fullmatch(DATE, "1879-03-14"))        # None    格式不同
# print(re.fullmatch(DATE, "March 14 1879"))     # None    少了逗號
```

#### **分組** 與 **具名分組**

加上括號，意思就是「整段都要配，但只拿括號裡的」。

```python
string = "Born: March 14, 1879"

m = re.search(r"([A-Z][a-z]+) (\d{1,2}), (\d{4})", string)

print(m.group())      # 'March 14, 1879'           整段
print(m.groups())     # ('March', '14', '1879')    每個分組
print(m.group(3))     # '1879'                     取第 3 組


# 正規表示式有分組時，findall 回傳的是「分組內容」（tuple 清單），非整段文字。
print(re.findall(r"([A-Z][a-z]+) (\d{1,2}), (\d{4})", "March 14, 1879 / June 01, 1926"))
# [('March', '14', '1879'), ('June', '01', '1926')]
```

```python
string = "tom.wu @ Tainan @ 1112"
    
re.findall(r"@ \w+", string)      # ['@ Tainan', '@ 1112']   @ 也一起帶走了
re.findall(r"@ (\w+)", string)    # ['Tainan', '1112']       ( ) 只帶走括號裡的
```

```python
string = "tom.wu @ Tainan @ 1112"

m = re.search(r"(\w+)\.(\w+)", string)  # <re.Match object; span=(0, 6), match='tom.wu'>

print(m.group())   # tom.wu  整段
print(m.group(1))  # tom     取第 1 組 
print(m.group(2))  # wu      取第 2 組
print(m.groups())  # ('tom', 'wu')  所有組合成一個 tuple
```

```python
string = "tom.wu @ Tainan @ 1112"

m = re.search(r"(\w+)\.(\w+) @ (\w+) @ (\d+)", string)

print(m.groups())    # ('tom', 'wu', 'Tainan', '1112')  所有組合成一個 tuple
print(m.group(3))    # 'Tainan'   ← 第 3 組是什麼？要回去數括號
print(m.group(4))    # '1112'     ← 第 4 組是什麼？要回去數括號
```

```python
# 具名分組 : 使用命名組來匹配字串，並將結果轉換為字典。

string = "tom.wu @ Tainan @ 1112"

m = re.search(
    r"(?P<first>\w+)\.(?P<last>\w+) @ (?P<city>\w+) @ (?P<number>\d+)",
    string
)

d = m.groupdict()
print(d)
# {'first': 'tom', 'last': 'wu', 'city': 'Tainan', 'number': '1112'}
```
